```python
import matplotlib.pyplot as plt
```

```python
names = ["Alice", "Bob", "Charlie", "Diana", "Evan"]
scores = [78, 85, 92, 88, 90]
```

```python
plt.plot(scores)
plt.xlabel("Student Index")
plt.ylabel("Score")
plt.title("Student Scores Line Plot")
plt.show()
```

```python
plt.bar(names, scores)
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Bar Chart")
plt.show()
```
